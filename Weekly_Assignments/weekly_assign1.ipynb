```python
"""
Take a year as input and print whether it is a leap year or not.
"""
```

```python
year = int(input("Enter a year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(" century Leap Year")
        else:
            print("Not Leap Year")
    else:
        print(" Leap Year")
else:
    print("Not Leap Year")
```

```python
"""
Take a positive integer n as input and print all prime numbers from 2 to n.
"""
```

```python
n = int(input("Enter a number: "))

for num in range(2, n + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num)
```

```python
"""
Keep accepting numbers from the user until they enter 0.
Then print the total count of numbers entered, their sum, and their average.
Do not include 0 in the calculations.
"""
```

```python

no = int(input("Enter a number: "))
count = 0
sum = 0


while no != 0:
    count = count + 1
    sum = sum + no

    no = int(input("Enter a number: "))

print("Count =", count)
print("Sum =", sum)

if count != 0:
    print("Average =", sum / count)
```

```python
"""
Take n as input and print a triangle pattern of n rows made of asterisks,
where row i has (2i - 1) stars centered appropriately.
"""
```

```python
n = int(input("Enter rows: "))

for i in range(1, n + 1):
    print(" " * (n - i), end="")
    print("*" * (2 * i - 1))
```

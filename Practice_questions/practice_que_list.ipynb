```python
"""
sum of even numbers
write  apython code that iterates through the following
list numbers = [12,7,23,48,15,66,9].
calculate and print the sum of all even numbers in the list.
""";
```

```python
numbers = [12,7,23,48,15,66,9]

sum = 0

for i in numbers:
    if i % 2 == 0:
        sum = sum + i

print("sum of even numbers is:",sum)
        
```

```python
"""
finding the maximum element
without using the builtin max() function ,write code to find and print
the largest number in the list values =[8,2,10,5,17,3,14]
""";
```

```python
value = [8,2,10,5,17,3,14]

largest = value[0]

for i in value:
    if i > largest:
        largest = i

print("largest element:",largest)
```

```python
"""
Count Occurrences
Given the list items = ['red', 'blue', 'green', 'red', 'yellow', 'blue', 'red'], 
write code to count how many times the string 'red' appears
in the list and print the count. Do not use the list.count() method.
""";
```

```python
items = ['red', 'blue', 'green', 'red', 'yellow', 'blue', 'red']

count = 0

for i in items:
    if i == 'red':
        count += 1

print(" count is:",count)
```

```python
"""
Remove All Occurrences of an Element
Given a list mixed_numbers = [1, 5, 2, 5, 3, 5, 4, 5], 
write code to remove all instances of the number 5 from the list. 
The final list should only contain [1, 2, 3, 4]. 
You must achieve this by iterating and creating a new list
or modifying the existing one carefully.
""";
```

```python
mixed_numbers = [1, 5, 2, 5, 3, 5, 4, 5]

my_list = []

for i in mixed_numbers:
    if i != 5:
        my_list.append(i)

print(my_list)
```

```python
"""
List Palindrome Check
Write code to determine if a given list is a palindrome
(reads the same forwards and backwards).
Print "Palindrome" if it is, and "Not a Palindrome" otherwise.
test with:
  list1 = [1,2,3,2,1]
  list2 = ['a','b','c','b','d']
  you are not allowed to use [::-1] for comparison.
""";
```

```python
list1 = [1, 2, 3, 2, 1]

new_list = []

for i in range(len(list1)-1, -1, -1):
    new_list.append(list1[i])

if list1 == new_list:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
list2 = ['a', 'b', 'c', 'b', 'd']

new_list = []

for i in range(len(list2)-1, -1, -1):
    new_list.append(list2[i])

if list2 == new_list:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
"""
Merge and Sort Two Lists (Manual Sort)
Given two lists:
list_a = [3, 1, 4]
list_b = [2, 5, 0]
Combine these two lists into a single list 
and then sort the combined list in ascending order without using the list.sort() method
or sorted() function.
You'll need to implement a basic sorting algorithm (e.g., bubble sort logic) using loops.
""";
```

```python
list_a = [3, 1, 4]
list_b = [2, 5, 0]

combine = list_a + list_b

for i in range(len(combine)):
    for j in range(len(combine)-1):

        if combine[j] > combine[j+1]:
            temp = combine[j]
            combine[j] = combine[j+1]
            combine[j+1] = temp

print(combine)
```

```python
"""
Flatten a Nested List
You are given a list that contains integers and other lists (only one level of nesting).
For example nested_list = [1, [2, 3], 4, [5, 6, 7], 8].
Write code to create a new flat list containing all the integers from nested_list.
The result should be [1, 2, 3, 4, 5, 6, 7, 8].
""";
```

```python
nested_list = [1, [2, 3], 4, [5, 6, 7], 8]

flat_list = []

for i in nested_list:
    if type(i) == list:
        flat_list = flat_list + i
    else:
        flat_list.append(i)

print(flat_list)
```

```python
"""
Find Common Elements
Given two lists, list_p = [1, 2, 3, 4, 5] and 
list_q = [4, 5, 6, 7, 8], write code to find and
print all elements that are common to both lists.
The order of common elements does not matter.
You are not allowed to use set operations.
""";
```

```python
list_p = [1, 2, 3, 4, 5]
list_q = [4, 5, 6, 7, 8]

common = []

for i in list_p:
    if i in list_q:
        common.append(i)

print(common)
```

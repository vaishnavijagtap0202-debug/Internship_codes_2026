```python
# create s function to find the factorial of a number
```

```python
def fac(n):
    fact = 1
    for i in range(1, n+1):
        fact = fact* i
    print("fact:",fact)
```

```python
fac(5)
```

```python
# typehinting => tell the type of parameters and return value and docstring  
```

```python
def sq(x:int)->int:
    """
    calculate the square of a number.

    Args:
        x(int):the number to square.
    returns:
        int: The square of a number.
    """
    res = x*x
    return res
```
